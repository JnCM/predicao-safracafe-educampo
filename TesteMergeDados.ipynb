```python
df_dataset_vcomarea = df_dataset_vcomarea.drop(['Municipio_id', 'altitude', 'Mesorregião'], axis=1) 
```

```python
df_dataset_vcomarea["Municipio"] = df_dataset_vcomarea["Municipio"].apply(lambda x: ''.join([c for c in unicodedata.normalize('NFKD', x.split(" - ")[0].replace("ç", "c")) if not unicodedata.combining(c)]))
df_dataset_vcomarea["Municipio"] = df_dataset_vcomarea["Municipio"].apply(lambda x: x.replace(" ", "_"))
```

```python
df_dataset_vcomarea["target_por_area"] = df_dataset_vcomarea["target"] / df_dataset_vcomarea["Área colhida (Hectares)"]
```

```python
df_dataset_vcomarea = df_dataset_vcomarea[df_dataset_vcomarea["Ano"] <= 2023]
```

```python
df_dadosclima['time'] = pd.to_datetime(df_dadosclima['time'])
df_dadosclima["Mes"] = df_dadosclima["time"].dt.month
df_dadosclima["Ano"] = df_dadosclima["time"].dt.year
```

```python
df_dadosclima = df_dadosclima.drop('time', axis=1) 
```

```python
df_dadosclima = df_dadosclima.loc[(df_dadosclima["Ano"] >= 2011) & (df_dadosclima["Ano"] <= 2023)]
# df_dadosclima = df_dadosclima[df_dadosclima["Ano"] >= 2011]
```

```python
df_dadosclima = df_dadosclima.drop('Base', axis=1) 
```

```python
missing_cities_df = df_dataset_vcomarea[~df_dataset_vcomarea['Municipio'].isin(df_dadosclima['Municipio'])]
len(missing_cities_df['Municipio'].unique())
```

```python
df_dadosclima = df_dadosclima.drop("WSDI", axis=1)
df_dadosclima = df_dadosclima.fillna(0)
df_dadosclima
```

```python
df_result = pd.DataFrame()
municipios = df_dataset_vcomarea["Municipio"].unique()
municipios_clima = df_dadosclima["Municipio"].unique()
anos = [2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023]
meses = [
    {6: "Junho"},{7: "Julho"},{8: "Agosto"},{9: "Setembro"},
    {10: "Outubro"},{11: "Novembro"},{12: "Dezembro"},{1: "Janeiro"},
    {2: "Fevereiro"},{3: "Março"},{4: "Abril"},{5: "Maio"}
]
i = 0
for municipio in municipios:
    if municipio in municipios_clima:
        for ano in anos:
            df_result.loc[i, "Municipio"] = municipio
            df_result.loc[i, "Ano"] = ano
            
            df_result.loc[i, "target_por_area"] = df_dataset_vcomarea.loc[
                (df_dataset_vcomarea["Municipio"] == municipio) &
                (df_dataset_vcomarea["Ano"] == ano), "target_por_area"
            ].iloc[0]

            colunas_clima = list(df_dadosclima.columns)
            for column in ["Municipio", "Mes", "Ano"]: colunas_clima.remove(column)
            
            for mes in meses:
                mes_num = list(mes.keys())[0]
                if mes_num >= 6:
                    df_clima = df_dadosclima.loc[
                        (df_dadosclima["Municipio"] == municipio) &
                        (df_dadosclima["Ano"] == ano-1) &
                        (df_dadosclima["Mes"] == mes_num)
                    ]

                    df_result.loc[i, f"preco_{mes_num}"] = df_preco.loc[
                        (df_preco["Ano"] == ano-1),
                        f"{list(mes.values())[0]}"
                    ].iloc[0]
                else:
                    df_clima = df_dadosclima.loc[
                        (df_dadosclima["Municipio"] == municipio) &
                        (df_dadosclima["Ano"] == ano) &
                        (df_dadosclima["Mes"] == mes_num)
                    ]

                    df_result.loc[i, f"preco_{mes_num}"] = df_preco.loc[
                        (df_preco["Ano"] == ano),
                        f"{list(mes.values())[0]}"
                    ].iloc[0]
                
                for column in colunas_clima:
                    df_result.loc[i, f"{column}_{mes_num}"] = df_clima[column].iloc[0]

            i += 1
```

```python
for mes in meses:
    mes_num = list(mes.keys())[0]
    df_result[f"preco_{mes_num}"] = df_result[f"preco_{mes_num}"].str.replace(',', '.', regex=False).astype('float')
```

```python
df_result.to_excel("Dados para correlação v3 - Xavier.xlsx", index=False)
```

```python
df_corr = pd.DataFrame()
```

```python
for ano in df_result["Ano"].unique():
    df_ano = df_result[df_result["Ano"] == ano]
    numeric_cols = df_ano.select_dtypes(include=['float64', 'int64']).columns
    corr_matrix = df_ano[numeric_cols].corr()
    df_corr.loc[:, ano] = corr_matrix["target"]
```

```python
for municipio in df_result["Municipio"].unique():
    df_municipio = df_result[df_result["Municipio"] == municipio]
    numeric_cols = df_municipio.select_dtypes(include=['float64', 'int64']).columns
    corr_matrix = df_municipio[numeric_cols].corr()
    df_corr.loc[:, municipio] = corr_matrix["target"]
```

```python
df_corr.fillna(0, inplace=True)
```

```python
df_corr.loc[:, "MEDIA"] = df_corr.mean(axis=1)
df_corr.loc[:, "DP"] = df_corr.std(axis=1)
```

```python
df_corr.to_excel("Correlação de variáveis por ano - Xavier.xlsx")
```

```python
colunas_clima = list(df_dadosclima.columns)
for column in ["Municipio", "Base", "Mes", "Ano"]: colunas_clima.remove(column)

variable_groups = dict()
for column in colunas_clima:
    variable_groups[column] = [f"{column}_{list(mes.keys())[0]}" for mes in meses]
variable_groups["preco"] = [f"preco_{list(mes.keys())[0]}" for mes in meses]

fig = go.Figure()

all_vars_in_order = [var for group in variable_groups.values() for var in group]
unique_variables = sorted(list(set(all_vars_in_order)))

for var in unique_variables:
    fig.add_trace(
        go.Bar(
            x=df_corr.columns.tolist(),
            y=df_corr.loc[var],
            name=var,
            visible=False
        )
    )


buttons = []
for group_name, vars_in_group in variable_groups.items():
    visibility = [trace.name in vars_in_group for trace in fig.data]
    
    buttons.append(
        dict(
            label=group_name,
            method='restyle',
            args=[{'visible': visibility}]
        )
    )

fig.update_layout(
    updatemenus=[
        dict(
            active=0,
            buttons=buttons,
            direction="down",
            pad={"r": 10, "t": 10},
            showactive=True,
            x=1.14,
            xanchor="right",
            y=1.25,
            yanchor="top"
        )
    ],
    width=1200,
    height=600,
    title_text="Correlação de variáveis por ano",
    xaxis_title="Ano",
    yaxis_title="Correlação",
    barmode='group',
    xaxis=dict(
        rangeslider=dict(
            visible=True
        ),
        type="category"
    )
)

initial_group = list(variable_groups.keys())[0]
for trace in fig.data:
    if trace.name in variable_groups[initial_group]:
        trace.visible = True

fig.show()
fig.write_html("Correlação de variáveis por ano - Xavier.html")
```
