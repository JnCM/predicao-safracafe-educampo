```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.io as pio
import unicodedata
```

```python
df_preco = pd.read_csv("tabela.csv", encoding="ISO-8859-1", sep="\t")
df_educampo = pd.read_excel("Dados Educampo - Safras Café.xlsx")
df_dataset = pd.read_csv("dataset_v17.csv", sep=",")
df_dadosclima = pd.read_csv("Índices Climáticos Anuais Municípios MG 1995-2024 - ERA5.csv", sep=",")
```

```python
df_preco
```

```python
df_dataset = df_dataset.drop('Municipio_id', axis=1) 
```

```python
df_dataset["Municipio"] = df_dataset["Municipio"].apply(lambda x: ''.join([c for c in unicodedata.normalize('NFKD', x.split(" - ")[0].replace("ç", "c")) if not unicodedata.combining(c)]))
df_dataset["Municipio"] = df_dataset["Municipio"].apply(lambda x: x.replace(" ", "_"))
```

```python
df_dadosclima['time'] = pd.to_datetime(df_dadosclima['time'])
df_dadosclima["Ano"] = df_dadosclima["time"].dt.year
```

```python
df_dadosclima = df_dadosclima.drop('time', axis=1) 
```

```python
df_dadosclima = df_dadosclima[df_dadosclima["Ano"] >= 2012]
df_dadosclima
```

```python
df_dataset
```

```python
merged_df = pd.merge(df_dadosclima, df_dataset, on=['Municipio', 'Ano'], how='inner')
```

```python
merged_df = merged_df[list(df_dadosclima.columns) + ["target"]]
merged_df['Ano'] = merged_df['Ano'].astype(int)
merged_df
```

```python
len(merged_df["Municipio"].unique())
```

```python
df_result = pd.DataFrame()
```

```python
for municipio in merged_df["Municipio"].unique():
    df_municipio = merged_df[merged_df["Municipio"] == municipio]
    numeric_cols = df_municipio.select_dtypes(include=['float64', 'int64']).columns
    corr_matrix = df_municipio[numeric_cols].corr()
    df_result.loc[:, municipio] = corr_matrix["target"]
```

```python
df_result.fillna(0, inplace=True)
```

```python
df_result.loc[:, "MEDIA"] = df_result.mean(axis=1)
df_result.loc[:, "DP"] = df_result.std(axis=1)
```

```python
df_result.to_excel("Correlação de variáveis climáticas com produção dos munícipios - ERA5.xlsx")
```

```python
numeric_cols = df_municipio.select_dtypes(include=['float64', 'int64']).columns
corr_matrix = df_municipio[numeric_cols].corr()
corr_matrix.dropna(axis=1, how='all', inplace=True)
corr_matrix.dropna(axis=0, how='all', inplace=True)
fig = px.imshow(
    corr_matrix,
    color_continuous_scale='RdBu_r',
    zmin=-1, zmax=1,
    labels=dict(color='Correlação'),
    title="Mapa de calor de correlação de variáveis",
    width=1200,
    height=1000
)
fig.show()
pio.write_html(fig, file="Correlação de Variáveis - Dados Xavier - Alfenas.html", auto_open=True)
```
