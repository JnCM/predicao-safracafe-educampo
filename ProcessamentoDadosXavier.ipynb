```python
import os
import pandas as pd
```

```python
df_faltantes = pd.DataFrame()
```

```python
for file in os.listdir("faltantes"):
    municipio = file.split("-MG")[0]
    df_municipio = pd.read_csv(f"faltantes/{file}", sep=",")
    df_municipio.loc[:, "Municipio"] = municipio
    df_municipio.loc[:, "Base"] = "Xavier"
    df_faltantes = pd.concat([df_faltantes, df_municipio])
```

```python
df_faltantes
```

```python
df_dadosxavier = pd.read_csv("dados_xavier/Índices Climáticos Mensais Municípios MG 1995-2024 - Xavier.csv", sep=",")
```

```python
df_dadosxavier
```

```python
df_dadosxavier = pd.concat([df_dadosxavier, df_faltantes])
```

```python
df_dadosxavier.to_csv("dados_xavier/Índices Climáticos Mensais Municípios MG 1995-2024 - Xavier v2.csv", sep=",", index=False)
```
