```python
variaveis = [
    "5_area_plantada_ha",
    "14_producao_sacas"
]
```

```python
df_cafe[variaveis]
```

```python
df_cafe['5_area_plantada_ha'] = df_cafe['5_area_plantada_ha'].str.replace('.', '').str.replace(',', '.').astype(float)
df_cafe['14_producao_sacas'] = df_cafe['14_producao_sacas'].str.replace('.', '').str.replace(',', '.').astype(float)
df_cafe['18_produtividade_sacas_ha'] = df_cafe['18_produtividade_sacas_ha'].str.replace('.', '').str.replace(',', '.').astype(float)
```

```python
df_safra_municipios = df_cafe.groupby(by=["municipio_fazenda", "regiao_sebrae", "safra"])[variaveis].mean().reset_index()
df_safra_municipios["18_produtividade_sacas_ha"] = df_safra_municipios["14_producao_sacas"] / df_safra_municipios["5_area_plantada_ha"]
df_safra_municipios
```

```python
df_safra_municipios.to_excel("Dados Educampo - Safras Café.xlsx", index=False)
```

```python
safras = [
    '2014/2015',
    '2015/2016', '2016/2017', '2017/2018', '2018/2019', '2019/2020',
    '2020/2021', '2021/2022', '2022/2023', '2023/2024', '2024/2025',
]
```

```python
fig = px.line(
    df_safra_municipios,
    x="safra",
    y="18_produtividade_sacas_ha",
    color="municipio_fazenda",
    markers=True,
    title="Produtividade por Município ao Longo dos Anos"
)

fig.update_layout(
    xaxis_title="Safra",
    yaxis_title="Produtividade (sacas/hectare)",
    legend_title="Município",
    hovermode="x unified"
)

fig.update_xaxes(categoryorder='array', categoryarray=safras)

fig.show()
```

```python
fig = px.line(
    df_safra_municipios,
    x="safra",
    y="14_producao_sacas",
    color="municipio_fazenda",
    markers=True,
    title="Produção por Município ao Longo dos Anos"
)

fig.update_layout(
    xaxis_title="Safra",
    yaxis_title="Produção (sacas)",
    legend_title="Município",
    hovermode="x unified"
)

fig.update_xaxes(categoryorder='array', categoryarray=safras)

fig.show()
```

```python
fig = px.line(
    df_safra_municipios,
    x="safra",
    y="5_area_plantada_ha",
    color="municipio_fazenda",
    markers=True,
    title="Área plantada por Município ao Longo dos Anos"
)

fig.update_layout(
    xaxis_title="Safra",
    yaxis_title="Área plantada (hectare)",
    legend_title="Município",
    hovermode="x unified"
)

fig.update_xaxes(categoryorder='array', categoryarray=safras)

fig.show()
```
