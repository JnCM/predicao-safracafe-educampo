```python
corr_matrix = df_cafe[numeric_cols].corr()
corr_matrix.dropna(axis=1, how='all', inplace=True)
corr_matrix.dropna(axis=0, how='all', inplace=True)
fig = px.imshow(
    corr_matrix,
    color_continuous_scale='RdBu_r',
    zmin=-1, zmax=1,
    labels=dict(color='Correlação'),
    title="Mapa de calor de correlação de variáveis",
    width=1200,
    height=1000
)
fig.show()
pio.write_html(fig, file="dados_educampo/Correlação de Variáveis.html", auto_open=True)
```
