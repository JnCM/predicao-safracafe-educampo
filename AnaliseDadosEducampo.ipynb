```python
numeric_cols = df_cafe.select_dtypes(include='number').columns
# Histograms
for i in range(0, len(numeric_cols), 10):
    df_cafe[numeric_cols[i:i+10]].hist(figsize=(14, 10), bins=30)
    plt.tight_layout()
    plt.show()
```

```python
# Filter numeric columns with enough non-NaN and non-constant values
numeric_cols = df_cafe.select_dtypes(include=['float64', 'int64'])
valid_cols = [col for col in numeric_cols.columns if df_cafe[col].nunique() > 1 and df_cafe[col].notna().sum() > 0]

for i in range(0, len(valid_cols), 10):
    # Choose how many variables to plot (e.g., first 9 valid ones)
    subset_cols = valid_cols[i:i+9]

    # Setup grid
    n = len(subset_cols)
    rows, cols = 3, 3  # Adjust based on number of variables
    fig, axes = plt.subplots(rows, cols, figsize=(18, 10))
    axes = axes.flatten()

    # Plot each variable in its own subplot
    for i, col in enumerate(subset_cols):
        sns.boxplot(y=df_cafe[col], ax=axes[i])
        axes[i].set_title(col)
        axes[i].set_xlabel("")
        axes[i].set_ylabel(col[0:10] + "...")

    # Hide any unused subplots
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
corr_matrix = df_cafe[numeric_cols].corr()
corr_matrix.dropna(axis=1, how='all', inplace=True)
corr_matrix.dropna(axis=0, how='all', inplace=True)
fig = px.imshow(
    corr_matrix,
    color_continuous_scale='RdBu_r',
    zmin=-1, zmax=1,
    labels=dict(color='Correlação'),
    title="Mapa de calor de correlação de variáveis",
    width=1200,
    height=1000
)
fig.show()
pio.write_html(fig, file="dados_educampo/Correlação de Variáveis.html", auto_open=True)
```
