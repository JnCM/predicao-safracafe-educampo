```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.io as pio
import plotly.graph_objects as go
import unicodedata
```

```python
df_dados_correlacao = pd.read_excel("Dados para correlação v2 - Xavier.xlsx")
```

```python
df_dados_correlacao
```

```python
numeric_cols = df_dados_correlacao.select_dtypes(include=['float64', 'int64']).columns
corr_matrix = df_dados_correlacao[numeric_cols].corr()
```

```python
corr_matrix
```

```python
corr_matrix.to_excel("Matriz de correlação v2 - Xavier.xlsx")
```

```python
df_corr_target = corr_matrix["target_por_area"]
df_corr_target.describe()
```

```python
q_75 = df_corr_target.quantile(0.75)
top_25 = df_corr_target[df_corr_target >= q_75]
top_25
```

```python
top_25.sort_values(ascending=False, inplace=True)
```

```python
top_25.to_excel("Top 25 variáveis mais significativas v2 - Xavier.xlsx")
```
